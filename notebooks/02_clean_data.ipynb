```python
import pandas as pd
from pathlib import Path

RAW = Path('../data/raw/csv')
CLEAN = Path('../data/clean')
CLEAN.mkdir(parents=True, exist_ok=True)

WINDOW_START = pd.Timestamp('2011-01-01', tz='UTC')
WINDOW_END   = pd.Timestamp('2022-01-01', tz='UTC')   # exclusive
```

```python
patients = pd.read_csv(RAW / 'patients.csv')

keep = ['Id', 'BIRTHDATE', 'DEATHDATE', 'MARITAL', 'RACE', 'ETHNICITY',
        'GENDER', 'CITY', 'STATE', 'COUNTY']
dropped = [c for c in patients.columns if c not in keep]
patients = patients[keep]

print('dropped:', dropped)
print(patients.shape)
```

```python
patients = patients.rename(columns={
    'Id': 'patient_id', 'BIRTHDATE': 'birthdate', 'DEATHDATE': 'deathdate',
    'MARITAL': 'marital', 'RACE': 'race', 'ETHNICITY': 'ethnicity',
    'GENDER': 'gender', 'CITY': 'city', 'STATE': 'state', 'COUNTY': 'county'})
patients.columns
```

```python
patients['birthdate'] = pd.to_datetime(patients['birthdate'], errors='coerce')
patients['deathdate'] = pd.to_datetime(patients['deathdate'], errors='coerce')

print('birthdate failed to parse:', patients['birthdate'].isna().sum())
print('deathdate missing (alive):', patients['deathdate'].isna().sum())
print('death before birth:', (patients['deathdate'] < patients['birthdate']).sum())
```

```python
for col in ['gender', 'marital', 'race', 'ethnicity']:
    print(col, patients[col].value_counts(dropna=False).to_dict())
```

```python
patients.to_csv(CLEAN / 'patients.csv', index=False, date_format='%Y-%m-%d')
print('written', len(patients), 'rows')
```

```python
enc = pd.read_csv(RAW / 'encounters.csv')
print(enc.shape)

keep = ['Id', 'START', 'STOP', 'PATIENT', 'PAYER', 'ENCOUNTERCLASS', 'DESCRIPTION',
        'BASE_ENCOUNTER_COST', 'TOTAL_CLAIM_COST', 'PAYER_COVERAGE', 'REASONDESCRIPTION']
enc = enc[keep].rename(columns={
    'Id': 'encounter_id', 'START': 'start_time', 'STOP': 'stop_time',
    'PATIENT': 'patient_id', 'PAYER': 'payer_id', 'ENCOUNTERCLASS': 'encounter_class',
    'DESCRIPTION': 'description', 'BASE_ENCOUNTER_COST': 'base_cost',
    'TOTAL_CLAIM_COST': 'total_claim_cost', 'PAYER_COVERAGE': 'payer_coverage',
    'REASONDESCRIPTION': 'reason_desc'})
```

```python
enc['start_time'] = pd.to_datetime(enc['start_time'], errors='coerce', utc=True)
enc['stop_time']  = pd.to_datetime(enc['stop_time'],  errors='coerce', utc=True)
print('start failed:', enc['start_time'].isna().sum(), '| stop missing:', enc['stop_time'].isna().sum())
```

```python
before = len(enc)
enc = enc[(enc['start_time'] >= WINDOW_START) & (enc['start_time'] < WINDOW_END)].copy()
print(f'kept {len(enc)} of {before} ({len(enc)/before:.1%}); excluded {before - len(enc)}')
```

```python
enc['los_days'] = (enc['stop_time'] - enc['start_time']).dt.total_seconds() / 86400

print('stop before start:', (enc['los_days'] < 0).sum())
print('zero-length:', (enc['los_days'] == 0).sum())
print('over 365 days:', (enc['los_days'] > 365).sum())
print(enc.groupby('encounter_class')['los_days'].describe().round(2))
```

```python
print('encounters whose patient is not in patients:',
      (~enc['patient_id'].isin(patients['patient_id'])).sum())
print('encounters with no payer:', enc['payer_id'].isna().sum())
print('duplicate encounter ids:', enc['encounter_id'].duplicated().sum())
```

```python
print(enc['encounter_class'].value_counts().to_dict())
print('reason_desc missing:', enc['reason_desc'].isna().sum(), 'of', len(enc))
```

```python
out = enc.copy()
out['start_time'] = out['start_time'].dt.strftime('%Y-%m-%d %H:%M:%S')
out['stop_time']  = out['stop_time'].dt.strftime('%Y-%m-%d %H:%M:%S')
out = out.drop(columns=['los_days'])        # MySQL computes this in the view
out.to_csv(CLEAN / 'encounters.csv', index=False)
print('written', len(out))
```

```python
cond = pd.read_csv(RAW / 'conditions.csv').rename(columns={
    'START': 'start_date', 'STOP': 'stop_date', 'PATIENT': 'patient_id',
    'ENCOUNTER': 'encounter_id', 'CODE': 'code', 'DESCRIPTION': 'description'})
cond['start_date'] = pd.to_datetime(cond['start_date'], errors='coerce')
cond['stop_date']  = pd.to_datetime(cond['stop_date'],  errors='coerce')

before = len(cond)
cond = cond[cond['encounter_id'].isin(enc['encounter_id'])].copy()
print(f'kept {len(cond)} of {before} conditions whose encounter is in the window')
print('stop_date missing (ongoing):', cond['stop_date'].isna().sum())
print('duplicate rows:', cond.duplicated().sum())

cond['description'] = cond['description'].str.strip()
cond['is_disorder'] = ~cond['description'].str.contains(r'\((?:finding|situation|person)\)?$', regex=True)
print(cond['is_disorder'].value_counts())
print(cond.loc[cond['is_disorder'], 'description'].value_counts().head(10))

cond['is_disorder'] = cond['is_disorder'].astype(int)
cond.to_csv(CLEAN / 'conditions.csv', index=False, date_format='%Y-%m-%d')
```

```python
pay = pd.read_csv(RAW / 'payers.csv')
pay = pay[['Id', 'NAME']].rename(columns={'Id': 'payer_id', 'NAME': 'name'})
print(pay)
pay.to_csv(CLEAN / 'payers.csv', index=False)
```
