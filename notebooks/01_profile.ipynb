```python
from pathlib import Path
import pandas as pd

# Locate the project root from the notebook's working directory.
cwd = Path.cwd()

if cwd.name == "notebooks":
    project_root = cwd.parent
else:
    project_root = cwd

# Find the CSV files without changing the raw data.
def find_csv(filename):
    matches = [
        project_root / "data" / "raw" / "csv" / filename,
        project_root / "data" / "raw" / filename,
    ]

    for path in matches:
        if path.exists():
            return path

    raise FileNotFoundError(
        f"Could not find {filename}. "
        f"Looked in: {[str(p) for p in matches]}"
    )


# Load the two remaining tables and encounters.
conditions = pd.read_csv(find_csv("conditions.csv"))
payers = pd.read_csv(find_csv("payers.csv"))
encounters = pd.read_csv(find_csv("encounters.csv"))

# Profile a table.
def profile_table(df, table_name):
    print("\n" + "=" * 65)
    print(f"TABLE: {table_name.upper()}")
    print("=" * 65)

    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]:,}")

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nMissing values:")
    print(df.isna().sum().to_string())

    print(f"\nDuplicate rows: {df.duplicated().sum():,}")

    if "Id" in df.columns:
        print(f"Duplicate IDs: {df['Id'].duplicated().sum():,}")

    print("\nData types:")
    print(df.dtypes.to_string())


# Profile conditions.
profile_table(conditions, "conditions")

print("\nCONDITION DESCRIPTIONS — value_counts()")
print(conditions["DESCRIPTION"].value_counts(dropna=False).to_string())

# Profile payers.
profile_table(payers, "payers")

print("\nPAYER NAMES — value_counts()")
print(payers["NAME"].value_counts(dropna=False).to_string())

# Examine the condition STOP column.
if "STOP" in conditions.columns:
    missing_stop = conditions["STOP"].isna().sum()

    print("\nCONDITION STOP ANALYSIS")
    print(f"Missing STOP: {missing_stop:,}")
    print(
        f"Missing STOP (%): "
        f"{conditions['STOP'].isna().mean() * 100:.2f}%"
    )

# Check the selected analysis window.
encounters["START"] = pd.to_datetime(
    encounters["START"],
    errors="coerce",
    utc=True
)

window_start = pd.Timestamp("2011-01-01", tz="UTC")
window_end = pd.Timestamp("2022-01-01", tz="UTC")

in_window = (
    encounters["START"].ge(window_start)
    & encounters["START"].lt(window_end)
)

print("\n" + "=" * 65)
print("ANALYSIS WINDOW: 2011–2021")
print("=" * 65)

print(f"Total encounters: {len(encounters):,}")
print(f"Encounters retained: {in_window.sum():,}")
print(f"Encounters excluded before 2011: "
      f"{encounters['START'].lt(window_start).sum():,}")
print(f"Encounters excluded from 2022 onward: "
      f"{encounters['START'].ge(window_end).sum():,}")
print(f"Encounters with invalid or missing START: "
      f"{encounters['START'].isna().sum():,}")
print(f"Retention rate: {in_window.mean() * 100:.2f}%")
```

```python
from pathlib import Path

# Profile the four project tables.
patients = pd.read_csv(find_csv("patients.csv"))

# Reuse the encounters DataFrame already loaded above.
tables = {
    "patients": patients,
    "encounters": encounters,
    "conditions": conditions,
    "payers": payers,
}

# Define primary keys and relationships.
table_notes = {
    "patients": {
        "key": "Id",
        "joins": "Referenced by encounters.PATIENT.",
    },
    "encounters": {
        "key": "Id",
        "joins": (
            "encounters.PATIENT → patients.Id; "
            "conditions.ENCOUNTER → encounters.Id; "
            "encounters.PAYER → payers.Id."
        ),
    },
    "conditions": {
        "key": "Id",
        "joins": "conditions.ENCOUNTER → encounters.Id.",
    },
    "payers": {
        "key": "Id",
        "joins": "Referenced by encounters.PAYER.",
    },
}

lines = [
    "# Hospital Patient Analytics Dashboard",
    "",
    "## Data Profile",
    "",
    "### Dataset provenance",
    "",
    "The dataset consists of synthetic healthcare records generated by "
    "Synthea. The records do not represent real patients or real hospital "
    "operations.",
    "",
    "### Table overview",
    "",
    "| Table | Rows | Columns | Duplicate rows | Duplicate IDs | Primary key |",
    "|---|---:|---:|---:|---:|---|",
]

for name, df in tables.items():
    duplicate_rows = int(df.duplicated().sum())

    if "Id" in df.columns:
        duplicate_ids = int(df["Id"].duplicated().sum())
        primary_key = "Id"
    else:
        duplicate_ids = None
        primary_key = "Not identified"

    duplicate_ids_text = (
        f"{duplicate_ids:,}" if duplicate_ids is not None else "N/A"
    )

    lines.append(
        f"| {name} | {len(df):,} | {df.shape[1]:,} | "
        f"{duplicate_rows:,} | {duplicate_ids_text} | {primary_key} |"
    )

lines += [
    "",
    "### Table relationships",
    "",
    "| Table | Key | Relationship / join |",
    "|---|---|---|",
]

for name, df in tables.items():
    note = table_notes[name]
    lines.append(
        f"| {name} | `{note['key']}` | {note['joins']} |"
    )

for name, df in tables.items():
    lines += [
        "",
        f"### {name.title()}",
        "",
        f"- Rows: {len(df):,}",
        f"- Columns: {df.shape[1]:,}",
        f"- Exact duplicate rows: {df.duplicated().sum():,}",
        "",
        "#### Missing values by column",
        "",
        "| Column | Missing values | Missing (%) |",
        "|---|---:|---:|",
    ]

    missing_counts = df.isna().sum()

    for column, count in missing_counts.items():
        percentage = count / len(df) * 100 if len(df) else 0
        lines.append(
            f"| `{column}` | {count:,} | {percentage:.2f}% |"
        )

    if "Id" in df.columns:
        lines += [
            "",
            f"- Duplicate IDs: {df['Id'].duplicated().sum():,}",
        ]

    if name == "conditions":
        lines += [
            "",
            "#### Most frequent condition descriptions",
            "",
            "| Description | Count |",
            "|---|---:|",
        ]

        counts = df["DESCRIPTION"].value_counts(dropna=False)

        for description, count in counts.items():
            safe_description = str(description).replace("|", "\\|")
            lines.append(f"| {safe_description} | {count:,} |")

        if "STOP" in df.columns:
            missing_stop = int(df["STOP"].isna().sum())
            missing_stop_pct = df["STOP"].isna().mean() * 100

            lines += [
                "",
                f"Missing `STOP` values: {missing_stop:,} "
                f"({missing_stop_pct:.2f}%).",
                "",
                "Missing STOP values may indicate ongoing simulated "
                "conditions. They will be retained as null unless a "
                "specific analysis requires another treatment.",
            ]

    if name == "payers":
        lines += [
            "",
            "#### Payer name distribution",
            "",
            "| Payer name | Count |",
            "|---|---:|",
        ]

        counts = df["NAME"].value_counts(dropna=False)

        for payer_name, count in counts.items():
            safe_name = str(payer_name).replace("|", "\\|")
            lines.append(f"| {safe_name} | {count:,} |")

# Document the chosen analysis window.
valid_start = encounters["START"].notna()
in_window = (
    encounters["START"].ge(window_start)
    & encounters["START"].lt(window_end)
)

before_2011 = encounters["START"].lt(window_start)
after_2021 = encounters["START"].ge(window_end)

lines += [
    "",
    "## Analysis window",
    "",
    "**Selected window: 2011-01-01 to 2021-12-31, inclusive.**",
    "",
    "The dashboard describes a defined period of simulated hospital "
    "operations rather than each patient's entire lifetime history. "
    "The full source tables will be retained in the database; the "
    "window will be applied in SQL views and Power BI.",
    "",
    f"- Total encounters: {len(encounters):,}",
    f"- Encounters retained in the window: {int(in_window.sum()):,}",
    f"- Encounters before 2011: {int(before_2011.sum()):,}",
    f"- Encounters from 2022 onward: {int(after_2021.sum()):,}",
    f"- Encounters with missing or invalid START: "
    f"{int((~valid_start).sum()):,}",
    "",
    "Data-quality log decision: analysis window: 2011–2021; "
    f"{int(before_2011.sum()):,} encounters before 2011 excluded "
    "because the dashboard describes recent operations, not lifetime "
    "history. The full tables remain available for other analyses.",
    "",
    "## Interpretation and limitations",
    "",
    "- The dataset is synthetic and should not be interpreted as "
    "evidence about a real hospital or population.",
    "- Missing patient DEATHDATE values are treated as indicating that "
    "the patient was alive at the end of the simulation.",
    "- Missing encounter REASONCODE values are retained as null and "
    "will not be used in dashboard indicators. The top-ten diagnosis "
    "chart will be based on the conditions table.",
    "- Missing condition STOP values are retained as null because "
    "they can represent ongoing simulated conditions.",
    "- Identifier-like patient fields will be assessed for removal "
    "during the cleaning stage.",
    "",
    "## Scope of this profile",
    "",
    "This document records initial profiling results. It does not "
    "represent completion of the cleaning process. Date validation, "
    "relationship validation, and other cleaning decisions will be "
    "documented separately.",
]

docs_dir = project_root / "docs"
docs_dir.mkdir(parents=True, exist_ok=True)

profile_path = docs_dir / "data_profile.md"
profile_path.write_text("\n".join(lines) + "\n", encoding="utf-8")

print(f"Created: {profile_path}")
```
